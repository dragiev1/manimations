```python
from manim import *
```

```python

class DerivativeDefinition(Scene):
    def construct(self):

        # 1. Setup the Coordinate System and Curve
        axes = Axes(x_range=[0, 10], y_range=[0, 2], x_length=8, y_length=3, axis_config={"include_tip": False})
        axes.to_corner(LEFT, buff=0.5).shift(RIGHT * 2)
        
        def f(x):
            return np.sin(x)
            
        graph = axes.plot(f, x_range=[0, 7.5], color=BLUE)
        graph_label = MathTex("f(x) = \\sin(x)", font_size=30).next_to(graph, UR, buff=0.2).shift(DOWN/4)
        
        self.play(FadeIn(axes), run_time=0.5)
        self.play(Create(graph), Write(graph_label))
        self.wait()

        # 2. Points and the Secant Line
        x_val = 1.5
        dx_tracker = ValueTracker(5.0) 
        
        def get_p1():
            return axes.c2p(x_val, f(x_val))
            
        def get_p2():
            return axes.c2p(x_val + dx_tracker.get_value(), f(x_val + dx_tracker.get_value()))
            
        dot1 = Dot(get_p1(), color=RED_D)
        dot2 = always_redraw(lambda: Dot(get_p2(), color=RED_D))
        
        p1_tex = MathTex("(x, f(x))", font_size=30).next_to(dot1, UP, buff=0.2)
        p2_tex = always_redraw(lambda: 
            MathTex("(x+\\Delta x, f(x+\\Delta x))", font_size=28)
            .next_to(dot2, UR, buff=0.2).shift(RIGHT/3.5)
)
        
        def get_secant():
            p1 = get_p1()
            p2 = get_p2()
            
            # Calculate the direction vector
            vec = p2 - p1
            vec_length = np.linalg.norm(vec)
            
            # If points are too close, use the tangent direction
            if vec_length < 0.01:
                # Use derivative for tangent direction
                h = 0.001
                slope = (f(x_val + h) - f(x_val)) / h
                direction = np.array([1, slope * (axes.y_length / axes.x_length) * (axes.x_range[1] - axes.x_range[0]) / (axes.y_range[1] - axes.y_range[0]), 0])
                direction = direction / np.linalg.norm(direction)
                line_length = 9  # Fixed length for tangent line
                return Line(p1 - direction * line_length/2, p1 + direction * line_length/2, color=PURPLE_A)
            else:
                # Normal secant line - extend it
                direction = vec / vec_length
                line_length = 9  # Fixed length
                return Line(p1 - direction * line_length/2, p1 + direction * line_length/2, color=PURPLE_A)
            
        secant = always_redraw(get_secant)
        
        self.play(Create(dot1), Write(p1_tex))
        self.play(Create(dot2), Write(p2_tex))
        self.play(Create(secant))
        self.wait()

        # 3. The Math Equations - positioned to avoid overlap
        secant_text = Tex(R"Slope of Secant Line =", font_size = 46).to_corner(UL).shift(DOWN/3 + RIGHT).set_stroke(color=WHITE, width=1).set_fill(opacity=0)
        eq1 = MathTex("\\frac{f(x+\\Delta x) - f(x)}{(x+\\Delta x) - x}").next_to(secant_text, RIGHT, buff=0.5)

        self.play(Write(secant_text))
        self.play(Write(eq1))
        self.wait()
        
        eq2 = MathTex("\\frac{f(x+\\Delta x) - f(x)}{\\Delta x}").next_to(secant_text, RIGHT, buff=0.5)
        
        self.play(Transform(eq1, eq2))
        self.wait()

        # 4. The Limit and the Animation
        tangent_text = Tex(R"Slope of Tangent Line =", font_size=46).to_corner(UL).shift(DOWN/3 + RIGHT).set_stroke(color=WHITE, width=1).set_fill(opacity=0)
        eq3 = MathTex("f'(x)", "=", "\\lim_{\\Delta x \\to 0}", "\\frac{f(x+\\Delta x) - f(x)}{\\Delta x}").next_to(tangent_text, DR)
                
        self.play(
            dx_tracker.animate.set_value(0.05),
            Transform(secant_text, tangent_text),
            Transform(eq1, eq3),
            run_time=3
        )

        derivative_text = Tex(R"Derivative of f =", font_size=46).to_corner(UL).shift(DOWN/3 + RIGHT).set_stroke(color=WHITE, width=1).set_fill(opacity=0)
        self.play(Transform(secant_text, derivative_text), eq1.animate.next_to(derivative_text), run_time=3)
        
        # P2 is P1 now.
        self.play(FadeOut(p2_tex, dot2))
        self.wait(2)
        self.play(FadeOut(axes, eq1, secant_text, graph_label, dot1, p1_tex, graph, secant))
```

```python
manim -pqh DerivativeDefinition
```
